### Step 2: Content Mix and Demographic Distribution Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

# Format split
formats = dashboard_payload['format_split']
axes[0].pie([formats['Movies'], formats['TV_Shows']], labels=['Movies', 'TV Shows'], colors=['#E50914', '#221F1F'], autopct='%1.1f%%', startangle=140)
axes[0].set_title('Catalog Content Format Proportion')

# Top producing countries
top_c_df = pd.DataFrame(dashboard_payload['top_countries'])
axes[1].barh(top_c_df['country'].head(8), top_c_df['total'].head(8), color='#3182CE', edgecolor='black')
axes[1].invert_yaxis()
axes[1].set_title('Top 8 Content Producing Nations')
axes[1].set_xlabel('Total Titles')
plt.tight_layout()
plt.show()
```
